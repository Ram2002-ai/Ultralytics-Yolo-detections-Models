```python
exercise_list=['Legextension','Legcurl','Squat','Deadlift','Benchpress','Overheadpress','Barbellrow','Bicepcurl','Tricepdip','Pushup','Pullup','Lunge','Plank','Burpee','Mountainclimber','Jumpingjack','Highknees','Buttkick','RussianTwist','Situp','Crunch','Legpress']
for exercise in exercise_list:
    safe_download(f"https://github.com/ultralytics/assets/releases/download/v0.0.0/{exercise}.demo.video.mp4")
```

```python
cap=cv2.VideoCapture(r'E:\ultralytics-yolo-main\Squat.demo.video.mp4')

# assert cap.isOpened(), #"Error opening video stream or file"

# video writer

w,h,fps=(int(cap.get(x)) for x in [cv2.CAP_PROP_FRAME_WIDTH, cv2.CAP_PROP_FRAME_HEIGHT, cv2.CAP_PROP_FPS])  

video_writer=cv2.VideoWriter('Squat.demo.video.out.mp4', cv2.VideoWriter_fourcc(*'mp4v'), fps, (w,h))

# Init AIGym
gym=solutions.AIGym(
    show=True,#displaye the frame
    kpts=[5,7,9,11,13] ,#keypoints index of person for monitoring specific exercise, for example, 5,7,9 are the keypoints index of left shoulder, left elbow and left wrist which are important for monitoring upper body exercises like benchpress, overheadpress, barbellrow, bicepcurl and tricepdip
    model='yolov8n-pose.pt' ,
    line_width=4,
    verbose=False
)

# process video frame by frame
while cap.isOpened():
    success,im0=cap.read()
    if not success:
        print("Video frame is empty - processing complete")
        break
    results=gym(im0) #monitir workouts on each frame
    video_writer.write(results.plot_img()) #write the processed frame to output video
video_writer.release() #release the video writer
cap.release() #release the video capture    
```

```python
cap = cv2.VideoCapture(r"E:\ultralytics-yolo-main\Pushup.demo.video.mp4")
# assert cap.isOpened() #"Error reading video file"

# Video writer
w, h, fps = (int(cap.get(x)) for x in (cv2.CAP_PROP_FRAME_WIDTH, cv2.CAP_PROP_FRAME_HEIGHT, cv2.CAP_PROP_FPS))
video_writer = cv2.VideoWriter("Pushups.demo.video.output.mp4", cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))

# Init AIGym
gym = solutions.AIGym(
    show=True,  # Display the frame
    kpts=[5, 7, 9],  # keypoints index of person for monitoring specific exercise, by default it's for pushup
    model="yolo11n-pose.pt",  # Path to the YOLO11 pose estimation model file
    line_width=4,  # Adjust the line width for bounding boxes and text display
    verbose=False,
)

# Process video
while cap.isOpened():
    success, im0 = cap.read()
    if not success:
        print("Video frame is empty or video processing has been successfully completed.")
        break
    results = gym(im0)  # monitor workouts on each frame
    video_writer.write(results.plot_img())  # write the output frame in file.

video_writer.release()
```
